```python
# Import necessary libraries
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Load the breast cancer dataset
data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target

# Display the first few rows
df.head()
```

```python
# How many features (columns), what are the names of columns?
print("=== Number of features ===")
# There are 30 feature columns + 1 target column = 31 total
# The feature names are stored in data.feature_names
print(f"Number of feature columns: {len(data.feature_names)}")
print(f"Number of total columns (including target): {df.shape[1]}")
print("\nColumn names:")
for i, col in enumerate(df.columns, 1):
    print(f"  {i}. {col}")
```

```python
# Data types of columns
print("\n=== Data types ===")
print(df.dtypes)
# Or a summary:
print("\nData type counts:")
print(df.dtypes.value_counts())
```

```python
# Is there any missing values?
print("\n=== Missing values ===")
print(f"Total missing values: {df.isnull().sum().sum()}")
print("\nMissing values per column:")
print(df.isnull().sum())
```

```python
# What are the unique values of the target column?
print("\n=== Unique target values ===")
print(f"Unique values: {df['target'].unique()}")
print(f"Value counts:\n{df['target'].value_counts()}")
```

```python
# Separate features and target variable
X = df.drop(columns='target')
y = df['target']

# Standardize the features to similar ranges
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Print first item of X and X_scaled to compare the values
print("=== First row of X (original) ===")
print(X.iloc[0])
print("\n=== First row of X_scaled (standardized) ===")
print(X_scaled[0])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.3, random_state=42
)
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
# Make predictions
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")
```

```python
# Additional evaluation metrics
print("\n=== Confusion Matrix ===")
print(confusion_matrix(y_test, y_pred))

print("\n=== Classification Report ===")
print(classification_report(y_test, y_pred, target_names=['malignant', 'benign']))
```

```python
from sklearn.neighbors import KNeighborsClassifier

# Train KNN
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)

# Predict and evaluate
y_pred_knn = knn_model.predict(X_test)
accuracy_knn = accuracy_score(y_test, y_pred_knn)
print(f"KNN Accuracy: {accuracy_knn:.2f}")
```

```python
from sklearn.tree import DecisionTreeClassifier

# Train Decision Tree
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)

# Predict and evaluate
y_pred_dt = dt_model.predict(X_test)
accuracy_dt = accuracy_score(y_test, y_pred_dt)
print(f"Decision Tree Accuracy: {accuracy_dt:.2f}")
```

```python
print("\n=== Model Comparison ===")
print(f"Logistic Regression : {accuracy_score(y_test, y_pred):.4f}")
print(f"K-Nearest Neighbors : {accuracy_knn:.4f}")
print(f"Decision Tree       : {accuracy_dt:.4f}")
```

```python
from sklearn.datasets import load_wine

wine = load_wine()
df_wine = pd.DataFrame(wine.data, columns=wine.feature_names)
df_wine['target'] = wine.target

print("=== Wine Dataset ===")
print(f"Shape: {df_wine.shape}")           # (178, 14) → 13 features + target
print(f"Target classes: {wine.target_names}")  # 3 classes: class_0, class_1, class_2
print(f"Missing values: {df_wine.isnull().sum().sum()}")

# Preprocess
X_w = df_wine.drop(columns='target')
y_w = df_wine['target']

# Split (do this BEFORE scaling — best practice)
X_train_w, X_test_w, y_train_w, y_test_w = train_test_split(
    X_w, y_w, test_size=0.3, random_state=42
)

# Scale (fit on train, transform on both — best practice)
scaler_w = StandardScaler()
X_train_w = scaler_w.fit_transform(X_train_w)
X_test_w = scaler_w.transform(X_test_w)

# Logistic Regression
lr_w = LogisticRegression(max_iter=1000)
lr_w.fit(X_train_w, y_train_w)
acc_lr_w = accuracy_score(y_test_w, lr_w.predict(X_test_w))
print(f"\nLogistic Regression on Wine: {acc_lr_w:.4f}")

# KNN
knn_w = KNeighborsClassifier(n_neighbors=5)
knn_w.fit(X_train_w, y_train_w)
acc_knn_w = accuracy_score(y_test_w, knn_w.predict(X_test_w))
print(f"KNN on Wine:               {acc_knn_w:.4f}")

# Decision Tree
dt_w = DecisionTreeClassifier(random_state=42)
dt_w.fit(X_train_w, y_train_w)
acc_dt_w = accuracy_score(y_test_w, dt_w.predict(X_test_w))
print(f"Decision Tree on Wine:     {acc_dt_w:.4f}")
```

```python
from sklearn.datasets import load_diabetes

diabetes = load_diabetes()
df_dia = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
df_dia['target'] = diabetes.target

print("=== Diabetes Dataset ===")
print(f"Shape: {df_dia.shape}")        # (442, 11) → 10 features + target
print(f"Target range: {diabetes.target.min()} to {diabetes.target.max()}")
print(f"Missing values: {df_dia.isnull().sum().sum()}")
```

```python
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
import numpy as np

X_d = df_dia.drop(columns='target')
y_d = df_dia['target']

X_train_d, X_test_d, y_train_d, y_test_d = train_test_split(
    X_d, y_d, test_size=0.3, random_state=42
)

scaler_d = StandardScaler()
X_train_d = scaler_d.fit_transform(X_train_d)
X_test_d = scaler_d.transform(X_test_d)

def evaluate_regression(name, model):
    model.fit(X_train_d, y_train_d)
    pred = model.predict(X_test_d)
    rmse = np.sqrt(mean_squared_error(y_test_d, pred))
    r2 = r2_score(y_test_d, pred)
    print(f"{name:25s} RMSE={rmse:.2f}  R²={r2:.4f}")

evaluate_regression("Linear Regression", LinearRegression())
evaluate_regression("Ridge Regression", Ridge(alpha=1.0))
evaluate_regression("Decision Tree Reg.", DecisionTreeRegressor(random_state=42))
evaluate_regression("KNN Regressor", KNeighborsRegressor(n_neighbors=5))
```
